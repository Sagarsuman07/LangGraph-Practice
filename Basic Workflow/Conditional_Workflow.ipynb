```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict, Literal
```

```python
class QEquationState(TypedDict):
    a: int
    b: int
    c: int

    equation: str
    discriminant: float
    result: str
```

```python
def show_equation(state: QEquationState):
    eq=f"{state["a"]}X2+{state["b"]}X+{state["c"]}"
    return {"equation": eq}

def calculate_discriminant(state:QEquationState):
    disc=state["b"]**2 - (4*state["a"]*state["c"])
    return {"discriminant": disc}

def real_roots(state: QEquationState):
    root1= (-state["b"] + state["discriminant"]**0.5)/(2*state["a"])
    root2= (-state["b"] - state["discriminant"]**0.5)/(2*state["a"])

    res=f"Real roots are {root1} and {root2}"
    return {"result": res}

def repeated_roots(state: QEquationState):
    root= (-state["b"])/(2*state["a"])

    res= f"Only repeating root is {root}"
    return {"result": res}

def no_real_roots(state: QEquationState):
    res= "No real roots"
    return {"result": res}



def check_condition(state: QEquationState)-> Literal["real_roots","repeated_roots","no_real_roots"] :
    if state["discriminant"] > 0:
        return "real_roots"
    elif state["discriminant"] == 0:
        return "repeated_roots"
    else:
        return "no_real_roots"
```

```python
mygraph=StateGraph(QEquationState)

mygraph.add_node("show_equation", show_equation)
mygraph.add_node("calculate_discriminant", calculate_discriminant)
mygraph.add_node("real_roots", real_roots)
mygraph.add_node("repeated_roots", repeated_roots)
mygraph.add_node("no_real_roots", no_real_roots)


mygraph.add_edge(START, "show_equation")
mygraph.add_edge("show_equation", "calculate_discriminant")

mygraph.add_conditional_edges("calculate_discriminant", check_condition)

mygraph.add_edge("real_roots", END)
mygraph.add_edge("repeated_roots", END)
mygraph.add_edge("no_real_roots", END)

workflow=mygraph.compile()
```

```python
initial_state={
    "a": 2,
    "b": 7,
    "c": 2
}

final_state=workflow.invoke(initial_state)
print(final_state)

```
